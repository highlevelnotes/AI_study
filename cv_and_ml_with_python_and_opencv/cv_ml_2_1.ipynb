```python
from pathlib import Path
import numpy as np
import cv2


img_path = Path('./img14.jpg')
img1 = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)


img1 = cv2.resize(img1, (320, 240))


methods = [cv2.THRESH_BINARY,
           cv2.THRESH_BINARY_INV,
           cv2.THRESH_TRUNC,
           cv2.THRESH_TOZERO,
           cv2.THRESH_TOZERO_INV,
           cv2.THRESH_OTSU,
           cv2.THRESH_TRIANGLE,
           cv2.ADAPTIVE_THRESH_MEAN_C,
           cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
           cv2.ADAPTIVE_THRESH_MEAN_C,
           cv2.ADAPTIVE_THRESH_GAUSSIAN_C]
thres = 70
maxVal = 255
ress = []

for i in range(0, 7):
    ret, res = cv2.threshold(img1, thres, maxVal, methods[i])
    ress.append(res)

ress.append(cv2.adaptiveThreshold(img1, maxVal, methods[7], methods[0], 11, 0))
ress.append(cv2.adaptiveThreshold(img1, maxVal, methods[8], methods[0], 11, 0))
ress.append(cv2.adaptiveThreshold(img1, maxVal, methods[9], methods[0], 61, 0))
ress.append(cv2.adaptiveThreshold(img1, maxVal, methods[10], methods[0], 61, 0))


displays = [('input1', img1), ('res1', ress[0]), ('res2', ress[1]), ('res3', ress[2]),
            ('res4', ress[3]), ('res5', ress[4]), ('res6', ress[5]), ('res7', ress[6]),
            ('res8', ress[7]), ('res9', ress[8]), ('res10', ress[9]), ('res11', ress[10])]

for (name, out) in displays:
    cv2.imshow(name, out)

cv2.waitKey(0)
cv2.destroyAllWindows()


save_dir = Path('./code_res_imgs/c2_threshold')
save_dir.mkdir(parents=True, exist_ok=True)

for (name, out) in displays:
    output_path = save_dir / f'{name}.jpg'
    cv2.imwrite(str(output_path), out)
```

```python
from pathlib import Path
import cv2

img1 = cv2.imread('./img7.jpg', cv2.IMREAD_UNCHANGED)
img1 = cv2.resize(img1, (320, 240))

res1 = cv2.cvtColor(img1, cv2.COLOR_BGR2HSV)

res1_split = list(cv2.split(res1))
res1_split[2] = cv2.add(res1_split[2], 100)
res1_merge = cv2.merge(res1_split)
res1_merge = cv2.cvtColor(res1_merge, cv2.COLOR_HSV2BGR)

displays = [
    ('img1', img1),
    ('res1', res1),
    ('res1_split_0', res1_split[0]),
    ('res1_split_1', res1_split[1]),
    ('res1_split_2', res1_split[2]),
    ('res1_merge', res1_merge)
]

for (name, out) in displays:
    cv2.imshow(name, out)

cv2.waitKey(0)
cv2.destroyAllWindows()

save_dir = Path('./code_res_imgs/c2_merge')
save_dir.mkdir(parents=True, exist_ok=True)

for (name, out) in displays:
    output_path = save_dir / f'{name}.jpg'
    cv2.imwrite(str(output_path), out)
```
