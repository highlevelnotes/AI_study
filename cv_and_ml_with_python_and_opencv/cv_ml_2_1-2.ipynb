```python
from pathlib import Path
import numpy as np
import cv2


img_path = Path('./img14.jpg')
img1 = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)


img1 = cv2.resize(img1, (320, 240))


methods = [cv2.THRESH_BINARY,
           cv2.THRESH_BINARY_INV,
           cv2.THRESH_TRUNC,
           cv2.THRESH_TOZERO,
           cv2.THRESH_TOZERO_INV,
           cv2.THRESH_OTSU,
           cv2.THRESH_TRIANGLE,
           cv2.ADAPTIVE_THRESH_MEAN_C,
           cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
           cv2.ADAPTIVE_THRESH_MEAN_C,
           cv2.ADAPTIVE_THRESH_GAUSSIAN_C]
thres = 70
maxVal = 255
ress = []

for i in range(0, 7):
    ret, res = cv2.threshold(img1, thres, maxVal, methods[i])
    ress.append(res)

ress.append(cv2.adaptiveThreshold(img1, maxVal, methods[7], methods[0], 11, 0))
ress.append(cv2.adaptiveThreshold(img1, maxVal, methods[8], methods[0], 11, 0))
ress.append(cv2.adaptiveThreshold(img1, maxVal, methods[9], methods[0], 61, 0))
ress.append(cv2.adaptiveThreshold(img1, maxVal, methods[10], methods[0], 61, 0))


displays = [('input1', img1), ('res1', ress[0]), ('res2', ress[1]), ('res3', ress[2]),
            ('res4', ress[3]), ('res5', ress[4]), ('res6', ress[5]), ('res7', ress[6]),
            ('res8', ress[7]), ('res9', ress[8]), ('res10', ress[9]), ('res11', ress[10])]

for (name, out) in displays:
    cv2.imshow(name, out)

cv2.waitKey(0)
cv2.destroyAllWindows()


save_dir = Path('./code_res_imgs/c2_threshold')
save_dir.mkdir(parents=True, exist_ok=True)

for (name, out) in displays:
    output_path = save_dir / f'{name}.jpg'
    cv2.imwrite(str(output_path), out)
```

```python
from pathlib import Path
import cv2

img1 = cv2.imread('./img7.jpg', cv2.IMREAD_UNCHANGED)
img1 = cv2.resize(img1, (320, 240))

res1 = cv2.cvtColor(img1, cv2.COLOR_BGR2HSV)

res1_split = list(cv2.split(res1))
res1_split[2] = cv2.add(res1_split[2], 100)
res1_merge = cv2.merge(res1_split)
res1_merge = cv2.cvtColor(res1_merge, cv2.COLOR_HSV2BGR)

displays = [
    ('img1', img1),
    ('res1', res1),
    ('res1_split_0', res1_split[0]),
    ('res1_split_1', res1_split[1]),
    ('res1_split_2', res1_split[2]),
    ('res1_merge', res1_merge)
]

for (name, out) in displays:
    cv2.imshow(name, out)

cv2.waitKey(0)
cv2.destroyAllWindows()

save_dir = Path('./code_res_imgs/c2_merge')
save_dir.mkdir(parents=True, exist_ok=True)

for (name, out) in displays:
    output_path = save_dir / f'{name}.jpg'
    cv2.imwrite(str(output_path), out)
```

```python
from pathlib import Path
import cv2
import numpy as np

img1 = cv2.imread('./image1.jpg', cv2.IMREAD_UNCHANGED)
img2 = cv2.imread('./img2.jpg', cv2.IMREAD_UNCHANGED)
img1 = cv2.resize(img1, (320, 240))
img2 = cv2.resize(img2, (320, 240))

mask = np.full(shape=img1.shape, fill_value=0, dtype=np.uint8)
h, w = img1.shape[:2]
x = (int)(w/2) - 60
y = (int)(h/2) - 60
cv2.rectangle(mask, (x, y), (x+120, y+120), (255, 255, 255), -1)

ress = []
ress.append(cv2.add(img1, img2))
ress.append(cv2.addWeighted(img1, 0.5, img2, 0.5, 0))
ress.append(cv2.subtract(img1, img2))
ress.append(cv2.absdiff(img1, img2))
ress.append(cv2.bitwise_not(img1))
ress.append(cv2.bitwise_and(img1, mask))


displays = [
    ('img1', img1),
    ('img2', img2),
    ('add', ress[0]),
    ('subtract', ress[1]),
    ('diff', ress[2]),
    ('bitnot', ress[3]),
    ('bitand', ress[4]),
]

for (name, out) in displays:
    cv2.imshow(name, out)

cv2.waitKey(0)
cv2.destroyAllWindows()

save_dir = Path('./code_res_imgs/c2_arithmeticLogical')
save_dir.mkdir(parents=True, exist_ok=True)

for (name, out) in displays:
    output_path = save_dir / f'{name}.jpg'
    cv2.imwrite(str(output_path), out)
```

```python
from pathlib import Path
import cv2
import numpy as np

img1 = cv2.imread('./image2.jfif', cv2.IMREAD_UNCHANGED)

v = np.full(shape=img1.shape, fill_value=100, dtype=np.uint8)
v_n = np.full(shape=img1.shape, fill_value=255, dtype=np.uint8)

ress = []
ress.append(np.uint8(img1 + v))
ress.append(cv2.add(img1, v))
ress.append(cv2.subtract(v_n, img1))

displays = [
    ('img1', img1),
    ('add', ress[0]),
    ('포화', ress[1]),
    ('subtact', ress[2])
]

for (name, out) in displays:
    cv2.imshow(name, out)

cv2.waitKey(0)
cv2.destroyAllWindows()

save_dir = Path('./code_res_imgs/c2_brightadjustment')
save_dir.mkdir(parents=True, exist_ok=True)

for (name, out) in displays:
    output_path = save_dir / f'{name}.jpg'
    cv2.imwrite(str(output_path), out)
```

```python
from pathlib import Path
import cv2
import numpy as np

img1 = cv2.imread('./image2.jfif', cv2.IMREAD_UNCHANGED)

multi_lut = np.full(shape=[256], fill_value=0, dtype=np.uint8)
log_lut = np.full(shape=[256], fill_value=0, dtype=np.uint8)
invol_lut = np.full(shape=[256], fill_value=0, dtype=np.uint8)
sel_lut = np.full(shape=[256], fill_value=0, dtype=np.uint8)
multi_v = 2
gamma1 = 0.1
gamma2 = 0.6
thres1 = 5
thres2 = 100
max_v_log = 255 / np.log(1 + 255)
max_v_invol = 255 / np.power(255, gamma1)
max_v_sel = 100 / np.power(thres2, gamma2)
for i in range(256):
    val = i * multi_v
    if val > 255:
        val = 255
    multi_lut[i] = val
    log_lut[i] = np.round(max_v_log * np.log(1+i))
    invol_lut[i] = np.round(max_v_invol * np.power(i, gamma1))
    if i < thres1: sel_lut[i] = i
    elif i > thres2: sel_lut[i] = i
    else: sel_lut[i] = np.round(max_v_sel * np.power(i, gamma2))

ress = []
ress.append(cv2.LUT(img1, multi_lut))
ress.append(cv2.LUT(img1, log_lut))
ress.append(cv2.LUT(img1, invol_lut))
ress.append(cv2.LUT(img1, sel_lut))

displays = [
    ('img1', img1),
    ('multilut', ress[0]),
    ('loglut', ress[1]),
    ('invollut', ress[2]),
    ('sellut', ress[3])
]

for (name, out) in displays:
    cv2.imshow(name, out)

cv2.waitKey(0)
cv2.destroyAllWindows()

save_dir = Path('./code_res_imgs/c2_contrast')
save_dir.mkdir(parents=True, exist_ok=True)

for (name, out) in displays:
    output_path = save_dir / f'{name}.jpg'
    cv2.imwrite(str(output_path), out)
```
