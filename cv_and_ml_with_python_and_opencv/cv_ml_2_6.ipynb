```python
import cv2
from pathlib import Path
import numpy as np

img1 = cv2.imread('./image3.jpg', cv2.IMREAD_GRAYSCALE)

img1_blue = cv2.GaussianBlur(img1, (3, 3), 0)
res1 = cv2.Sobel(img1, cv2.FILTER_SCHARR, 1, 0, ksize=3)
res2 = cv2.Scharr(img1_blue, cv2.CV_32FC1, 0, 1)
res3 = cv2.Laplacian(img1_blue, cv2.CV_32FC1)
res4 = cv2.Canny(img1, 50, 200, apertureSize=5, L2gradient=True)

displays = [
    ('1', img1),
    ('2', res1),
    ('3', res2),
    ('4', res3),
    ('5', res4)
]

for (name, out) in displays:
    cv2.imshow(name, out)

cv2.waitKey(0)
cv2.destroyAllWindows()

save_dir = Path('./code_res_imgs/c2_edge')
save_dir.mkdir(parents=True, exist_ok=True)

for (name, out) in displays:
    output_path = save_dir / f'{name}.jpg'
    cv2.imwrite(str(output_path), out)
```

```python
img1 = cv2.imread('./image3.jpg', cv2.IMREAD_GRAYSCALE)

img1_edge = cv2.Canny(img1, 50, 150, apertureSize=3)
lines = cv2.HoughLines(img1_edge, 2, np.pi/180, 100)
linesP = cv2.HoughLinesP(img1_edge, 2, np.pi/180, 50, minLineLength=1, maxLineGap=100)

circles = cv2.HoughCircles(img1, cv2.HOUGH_GRADIENT, 1, 20, param1=50, param2=20, minRadius=30, maxRadius=50)

img1_color1 = cv2.cvtColor(img1, cv2.COLOR_GRAY2BGR)
if lines.any() != None:
    for line in lines:
        rho, theta = line[0]
        a = np.cos(theta)
        b = np.sin(theta)
        x0 = a * rho
        y0 = b * rho
        x1 = int(x0 + 1000 * (-b))
        y1 = int(y0 + 1000 * a)
        x2 = int(x0 - 1000 * (-b))
        y2 = int(y0 - 1000 * a)
        cv2.line(img1_color1, (x1, y1), (x2, y2), (0, 0, 255), 2)
img1_color2 = cv2.cvtColor(img1, cv2.COLOR_GRAY2BGR)
if linesP.any() != None:
    for line in linesP:
        x1, y1, x2, y2 = line[0]
        cv2.line(img1_color2, (x1, y1), (x2, y2), (0, 255, 0), 2)
circles = np.uint16(np.around(circles))
img2_color1 = cv2.cvtColor(img1, cv2.COLOR_GRAY2BGR)
if circles.any() != None:
    for i in circles[0, :]:
        cv2.circle(img2_color1, (i[0], i[1]), i[2], (0, 0, 255), 2)

displays = [
    ('1', img1),
    ('2', img1_color1),
    ('3', img1_color2),
    ('4', img2_color1),
]

for (name, out) in displays:
    cv2.imshow(name, out)

cv2.waitKey(0)
cv2.destroyAllWindows()

save_dir = Path('./code_res_imgs/c2_houghTransform')
save_dir.mkdir(parents=True, exist_ok=True)

for (name, out) in displays:
    output_path = save_dir / f'{name}.jpg'
    cv2.imwrite(str(output_path), out)
```
