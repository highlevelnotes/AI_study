```python
'''
    Example20.py
    함수
'''

def fibonacci(num):
    a, b = 0, 1
    while a < num:
        print(a)
        a, b = b, a+b
    print('end')

fibonacci(10)
```

```python
'''
    Example21.py
    함수
'''

def fibonacci(num):
    series = []
    a, b = 0, 1
    while a < num:
        series.append(a)
        a, b = b, a+b
    return series

result = fibonacci(10)
print(result)
```

```python
'''
    Example23.py
'''

def login(Message, times=3, alert='It\'s wrong ID'):
    while True:
        id = input(Message)
        if id in ('User', 'user', 'USER'):
            return True
        times = times - 1
        if times <= 0:
            print('invalid user response')
            return False
        print(alert)

login('아이디를 입력해주세요')
login('Input your ID', 2)
login('YOUR ID', 2, 'Try to type your ID')
```

```python
'''
    Example24.py
'''

def caffe(beverage, *arguments, **keywords):
    print('Do you have any', beverage, '?')
    for arg in arguments:
        print(arg)
    print('*******')
    for kw in keywords:
        print(kw, ':', keywords[kw])

caffe('coffee', "It's yummy, sir.", "It's the rich coffee aroma, sir.", barista='Michael Palin', client='Hong Gil Dong', cup='Ceramic Cup')
```

```python
'''
    Example25.py
'''

def plus(numA, numB):
    return numA + numB

def minus(numA, numB):
    return numA - numB

def multiply(numA, numB):
    return numA * numB

def divide(numA, numB):
    if numB == 0:
        print('0으로 나눌 수 없음')
        return -1
    else:
        return numA / numB
    
num1 = float(input('1번 숫자를 입력하세요 : '))
op = input('사칙연산자(+, -, *, /) 선택하세요 : ')
num2 = float(input('2번 숫자를 입력하세요 : '))

if op == '+':
    print('%f + %f = %f'%(num1, num2, plus(num1, num2)))
elif op == '-':
    print('%f - %f = %f'%(num1, num2, minus(num1, num2)))
elif op == '*':
    print('%f * %f = %f'%(num1, num2, multiply(num1, num2)))
elif op == '/':
    print('%f / %f = %f'%(num1, num2, divide(num1, num2)))
else:
    print('연산자 오류')
```
