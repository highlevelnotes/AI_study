```python
'''
    Example1.py
    주석, 변수
    여러 줄 주석은 큰따옴표 3개로 감싼다.
'''
width = 3
height = 5
RectArea = width*height
TriangularArea = width*height/2
print(RectArea)
print(width, '*', height, '=', RectArea)
print('{0} : {1} * {2} / 2 = {3}'.format('TriangularArea', width, height, TriangularArea))

# 숫자
num = 13
result = num / 4
Quotient = num // 4
Remainder = num % 4
print(num, '을 4로 나눈 결과는', result)
print('{0} / 4의 몫은 {1}이고 나머지는 {2}이다'.format(num, Quotient, Remainder))
```

```python
'''
    Example4.py
    변수
'''

# 16진수, 8진수, 2진수 변수 초기화
a = 0xFF
b = 0o67
c = 0b1010

print(a, b, c)
print('%x, %o %s' %(a, b, bin(c)))

# 12에 대한 10진수, 16진수, 8진수 출력
print('%d, %x, %o' %(12, 12, 12))

# 12에 대한 이진수, 16진수, 8진수 출력
print('%s %s %s' %(bin(12), hex(12), oct(12)))
```

```python
'''
    Example16.py
'''

word = input('type a word :')
wordlength = len(word)

chk = -1
for i in range(wordlength):
    print(word[i])
    if word[i] != word[wordlength - i - 1]:
        chk = i
        break

if chk == -1:
    print('0')
```

```python
'''
    Example18.py
    함수
'''

inStr = input('숫자열을 입력하세요 : ')
print('')

for idxStr in range(0, len(inStr)):
    ch = inStr[idxStr]
    numDollar = int(ch)
    strDollar = ''
    for idxDollar in range(numDollar):
        strDollar += '$'
    
    print(numDollar, ':', strDollar)
```

```python
'''
    Example19.py
    엘리베이터
'''

TargetFloor = 0
PresentFloor = 1
while True:
    PresentFloor = int(input('현재 층을 입력하세요(1~15) :'))
    if PresentFloor < 1 or PresentFloor > 15:
        print('현재 층은 1~15층 사이로 입력해야 합니다')
    else:
        break

while True:
    TargetFloor = int(input('가고 싶은 층을 입력하세요(1~15) : '))
    if TargetFloor < 1 or TargetFloor > 15 or TargetFloor == PresentFloor:
        print('현재 층은 %d이며 가고 싶은 층은 현재 층을 제외하고 1~15층 사이로 입력해야 합니다'%PresentFloor)
    else:
        break

if PresentFloor > TargetFloor:
    inc = -1
else:
    inc = 1

for idx in range(PresentFloor, TargetFloor, inc):
    print('현재 층은 %d입니다.'%idx)
print('현재 층은 %d입니다.'%TargetFloor)
```
