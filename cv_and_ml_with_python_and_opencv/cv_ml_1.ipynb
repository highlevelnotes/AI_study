```python
'''
    Example1.py
    주석, 변수
    여러 줄 주석은 큰따옴표 3개로 감싼다.
'''
width = 3
height = 5
RectArea = width*height
TriangularArea = width*height/2
print(RectArea)
print(width, '*', height, '=', RectArea)
print('{0} : {1} * {2} / 2 = {3}'.format('TriangularArea', width, height, TriangularArea))

# 숫자
num = 13
result = num / 4
Quotient = num // 4
Remainder = num % 4
print(num, '을 4로 나눈 결과는', result)
print('{0} / 4의 몫은 {1}이고 나머지는 {2}이다'.format(num, Quotient, Remainder))
```

```python
'''
    Example2.py
    문자열, 리스트
    여러 줄 주석은 큰따옴표 3개로 감싼다.
'''

# 문자열
strings = 'It is string'
print(strings)
print('{0}'.format('It\'s string'))
strings = "Double and single quotation marks have the same result"
print(strings)
strings = 'Seoul city'
print(strings[0], strings[2])
print(strings[-1], strings[-2], strings[-3])
print(strings[0:5] + strings[-4:-1] + strings[-1])

print('c:\programs\newpycharm')
print(r'c:\programs\newpycharm')

# 리스트
numbers = [11, 12, 13, 14, 15]
print(numbers)
print(numbers[2])
print(numbers[2:4])

numbers[1], numbers[3] = 22, 24
print(numbers[:])

numbers[:] = []
print(numbers[:])

numbers = [1, 2, 3]
alphabet = ['a', 'b', 'c']
collaboration = [numbers, alphabet]
print(collaboration[1])
```

```python
'''
    Example4.py
    변수
'''

# 16진수, 8진수, 2진수 변수 초기화
a = 0xFF
b = 0o67
c = 0b1010

print(a, b, c)
print('%x, %o %s' %(a, b, bin(c)))

# 12에 대한 10진수, 16진수, 8진수 출력
print('%d, %x, %o' %(12, 12, 12))

# 12에 대한 이진수, 16진수, 8진수 출력
print('%s %s %s' %(bin(12), hex(12), oct(12)))
```
