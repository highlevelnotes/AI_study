```python
'''
    Example31.py
'''

class Parent:
    num = 0
    def __init__(self, num):
        self.num = num
        print('The constructor of the parent class was called.')

class Child(Parent):
    def __init__(self, num):
        super().__init__(num)
        print('The constructor of the child class was called')
    def DisplayValue(self):
        print('num :', self.num)

cd = Child(20)
cd.DisplayValue()
```

```python
'''
    Example32.py
    프라퍼티
'''

class Dog:
    def __init__(self):
        self.__ownernames = 'default name'
    
    def get_name(self):
        return self.__ownernames

    def set_name(self, name):
        self.__ownernames = name

myDog = Dog()
myDog.set_name('Marry')
print(myDog.get_name())

```

```python
'''
    Example33
    프라퍼티
'''

class Dog:
    def __init__(self):
        self.__ownernames = 'default name'
    
    @property
    def name(self):
        return self.__ownernames

    @name.setter
    def name(self, name):
        self.__ownernames = name

myDog = Dog()
myDog.name = 'Marry'
print(myDog.name)
```

```python
'''
    Example38.py
'''

fp = open('./names.txt', 'r', encoding='utf-8')

names = fp.readlines()
fp.close()

total_names = 0
count = 0

for name in names:
    total_names += 1
    name_strip = name.strip('\n')
    if len(name_strip) <= 5:
        print('%s - len : %d' % (name_strip, len(name_strip)))
        count += 1

print('The total number of names is %d, and there are %d names with less than 5 letters.' % (total_names, count))
```
