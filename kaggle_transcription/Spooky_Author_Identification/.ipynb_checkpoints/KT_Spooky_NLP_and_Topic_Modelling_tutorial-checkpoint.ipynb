### 훈련 데이터의 요약 통계

여기에서는 데이터에서 각 작가별 항목의 분포와 같은 기본적인 통계 정보를 시각화할 수 있다. 이를 위해, 유용한 Plot.ly 시각화 라이브러리를 사용하여 간단한 막대 그래프를 그릴 것이다. Plot.ly 코드가 필요하면 아래 셀을 열어보자.

```python
z = {'EAP' : 'Edgar Allen Poe', 'MWS' : 'Mary Shelley', 'HPL': 'HP Lovecraft'}
data = [go.Bar(
            x = train.author.map(z).unique(),
            y = train.author.value_counts().values,
            marker = dict(colorscale='Jet', 
                         color = train.author.value_counts().values),
            text='Text entries attributed to Author'
        )]

layout = go.Layout(
    title='Target variable distribution'
)

fig = go.Figure(data=data, layout=layout)

py.iplot(fig, filename='basic-bar')
```

```python
all_words = train['text'].str.split(expand=True).unstack().value_counts()
data = [go.Bar(
            x = all_words.index.values[2:50],
            y = all_words.values[2:50],
            marker=dict(colorscale='Jet',
                       color=all_words.values[2:100]),
            text='Word counts'
)]

layout = go.Layout(
    title='Top 50 (Uncleaned) Word frequencies in the training dataset'
)

fig = go.Figure(data=data, layout=layout)

py.iplot(fig, filename='basic-bar')
```

이 단어 빈도 그래프에서 나타나는 단어들에 대해 뭔가 이상한 점이 있는지 확인해보자. 이 단어들이 실제로 Mary Shelley가 그녀의 이야기에서 독자에게 전달하려는 주제와 개념에 대해 많은 정보를 제공하는지 말이다.

이 단어들은 너무 흔하게 발생하는 단어들로, 다른 곳에서도 어디서든 찾아볼 수 있는 단어들이다. 우리 세 명의 작가들의 공포 이야기나 소설뿐만 아니라, 신문, 어린이 책, 종교적 텍스트 등 거의 모든 영어 텍스트에서 찾아볼 수 있다. 따라서 우리가 해야 할 일은 먼저 데이터셋을 전처리하여 이렇게 흔히 나오는 단어들을 제거하는 방법을 찾는 것이다.

### 워드클라우드를 통해 각 작가의 작품을 시각화

자연어 처리 작업에서 데이터 과학자에게 매우 유용한 시각화 도구 중 하나는 "워드클라우드"를 그리는 것이다. 워드클라우드는 (이름에서 알 수 있듯이) 텍스트나 책을 구성하는 다양한 단어들이 모여 있는 이미지이며, 각 단어의 크기는 해당 단어가 텍스트에서 나타나는 빈도(즉, 그 단어가 등장하는 횟수)에 비례한다. 여기에서는 실제 책이나 텍스트 대신, "text" 열에서 단어들을 가져올 수 있다.

* 각 작가의 텍스트를 파이썬 리스트에 저장

우리는 Edgar Allen Poe, HP Lovecraft, Mary Shelley의 텍스트를 각각 저장하는 세 개의 파이썬 리스트를 다음과 같이 생성한다.

```python
eap = train[train.author=='EAP']['text'].values
hpl = train[train.author=='HPL']['text'].values
mws = train[train.author=='MWS']['text'].values
```

다음, 우리의 워드 클라우드를 만들기 위해 파이썬 모듈 'wordcloud'를 임포트한다.

```python
from wordcloud import WordCloud, STOPWORDS
```

하지만 일반적인 워드클라우드를 생성하는 것은 다소 지겹기 때문에, 관련있는 사진을 임포트하여 그 사진의 아웃라인 마스크로 사용해 우리의 워드 클라우드를 구성하는 기술을 소개하고 싶다. 그렇게 하면 자신이 선택한 사진이 작가들을 가장 대표하는 것으로 느끼는 것들이 된다:

1) 에드거 앨런 포의 까마귀

2) HP 러프크래프트의 크툴루 문어같은 것

3) 메리 셸리의 프랑켄슈타인


Kaggle에서 이미지를 불러오는 방법은 일종의 기능 해킹인데, 내 작업에 익숙한 독자들은 이 방법을 알 것이다. 먼저 사용하려는 이미지들의 Base64 인코딩을 추출한 후, 해당 인코딩을 사용하여 이미지를 다시 노트북에 복원한다. 아래 셀에는 내가 사용할 세 개의 이미지에 대한 Base64 인코딩이 포함되어 있지만, 텍스트로 긴 줄이 노트북을 오염시키지 않도록 숨겨놓았다. 인코딩을 보고 싶다면 셀을 펼쳐보면 된다.

```python
eap_64 = b''
hpl_64 = b''

mws_64 = b''
```

이미지를 복호화하기 위해, codecs 모듈을 임포트하고 그것을 이 노트북에 새로운 이미지로 저장한다. 일단 이미지가 저장되면, 다음과 같이 마스크로 그것을 간단히 로드할 수 있다:

```python
import codecs

f1 = open('eap.png', 'wb')
f1.write(codecs.decode(eap_64, 'base64'))
f1.close()
img1 = imread('eap.png')
```

```python
img1
```

```python
hcmask = img1

f2 = open('mws.png', 'wb')
f2.write(codecs.decode(mws_64, 'base64'))
f2.close()
img2 = imread('mws.png')
hcmask2 = img2

f3 = open('hpl.png', 'wb')
f3.write(codecs.decode(hpl_64, 'base64'))
f3.close()
img3 = imread('hpl.png')
hcmask3 = img3
```

마지막으로 워드 클라우드를 다음의 몇 줄을 통해 시각화한다(코드를 보기 위해 숨김해제 한다):

```python
plt.figure(figsize=(16, 13))
wc = WordCloud(background_color='black', max_words=10000, mask=hcmask3, stopwords=STOPWORDS, max_font_size=40)
wc.generate(' '.join(hpl))
plt.title('HP Lovecraft (Cthulhu-Squidy)', fontsize=20)

plt.imshow(wc.recolor(colormap='Pastel2', random_state=17), alpha=0.98)
plt.axis('off')
```

```python
plt.figure(figsize=(20, 18))

wc = WordCloud(background_color='black',
              max_words=10000,
              mask=hcmask,
              stopwords=STOPWORDS,
              max_font_size=40)
wc.generate(' '.join(eap))
plt.show('Edgar Allen Poe (The Raven)')
plt.imshow(wc.recolor(colormap='PuBu', random_state=17), alpha=0.9)
plt.axis('off')
```

```python
plt.figure(figsize=(20, 18))
wc = WordCloud(background_color='black',
              max_words=10000,
              mask=hcmask2,
              stopwords=STOPWORDS,
              max_font_size=40)
wc.generate(' '.join(mws))
plt.title("Mary Shelley (Frankenstein's Monster)", fontsize=18)
plt.imshow(wc.recolor(colormap='viridis', random_state=17), alpha=0.9)
plt.axis('off')
```

자, 이제 여기 세 개의 별도의 워드클라우드가 있다. 각각의 공포 작가들을 위한 것들이다. 바로 이 워드클라우드들에서 각 작가가 선호하는 단어들을 알 수 있다.

예를 들어, HP Lovecraft는 "dream", "time", "strange", "past", "ancient"와 같은 단어들을 자주 사용하며, 이는 그 작가가 유명한 주제들과 관련이 있는 단어들이다. 이러한 주제들은 숨겨진 정신과 운명과 우연의 비밀스러운 성질, 그리고 악명 높은 괴물인 Cthulhu와 그와 관련된 고대 종교와 의식들에 대한 이야기들이다.

반면, Mary Shelley의 경우, 그녀의 단어들은 본능적인 충동과 도덕성에 대한 주제로 집중되며, "friend", "fear", "hope", "spirit" 등과 같은 단어들이 보인다. 이는 "Frankenstein"과 같은 그녀의 작품에서 다루는 주제들과 관련이 있다.

그러나 워드클라우드를 보면 여전히 꽤 어울리지 않는 단어들이 몇 가지 보인다. 예를 들어, "us", "go", "he"와 같은 단어들이 텍스트 곳곳에서 자주 등장하는데, 이는 어디서든 흔히 볼 수 있는 단어들이다.

## 2. 자연어 처리


거의 모든 자연어 처리(Natural Language Processing, NLP) 작업에서는 (주제 모델링, 단어 클러스터링, 문서 텍스트 분류 등) 입력 원시 텍스트를 모델과 머신이 읽을 수 있는 형태로 변환하는 몇 가지 전처리 단계를 반드시 거쳐야 한다. 랜덤 포레스트 모델에 단어들이 가득 담긴 문단을 그대로 넣고, 그 문단이 어떤 작가의 작품인지 바로 예측할 수 있다고 기대할 수는 없다. 실제로, 텍스트 전처리는 다음과 같은 몇 가지 간단한 단계로 요약할 수 있다:

1. 토큰화 (Tokenization) - 텍스트를 개별적인 단어로 나누는 과정.


2. 불용어 제거 (Stopwords) - 너무 자주 등장하는 단어들을 제거한다. 이러한 단어들은 텍스트를 구별하는 데 도움이 되지 않기 때문이다. (또한, 매우 드물게 등장하는 단어들도 제거하는 것을 고려할 수 있다.)


3. 형태소 분석 (Stemming) - 단어의 변형형들을 동일한 기본 단어로 결합하여 여전히 같은 의미를 전달하도록 한다.


4. 벡터화 (Vectorization) - 텍스트를 벡터 형식으로 변환하는 과정. 가장 간단한 방법 중 하나는 유명한 Bag-of-Words 접근법인데, 이 방법은 각 문서나 텍스트마다 행렬을 생성하는 것이다. 가장 간단한 형태에서는 이 행렬이 단어 빈도(단어 수)를 저장하며, 원시 텍스트를 벡터화한다고도 불린다.

자연어 툴킷(NLTK): 자연어 처리 작업을 더 편리하게 하기 위해, 가장 유용한 툴킷 중 하나인 자연어 툴킷(Natural Language Toolkit, NLTK)을 소개하려 한다. 이 툴킷을 임포트하는 것은 매우 간단하다:

```python
import nltk
```

### 2a. 토큰화

토큰화(tokenization)라는 개념은 주어진 문서에서 문자 시퀀스(파이썬 문자열을 생각해보자)를 취하고, 그것을 개별적인 구성 요소들로 나누는 행위이다. 이 구성 요소들을 이 방법의 이름을 딴 "토큰"이라고 한다. 이를 간단히 말하자면, 문장에서의 단어 하나하나라고 생각할 수 있다. 단순하게 문자열에서 "split()" 메서드를 구현하여 공백을 기준으로 파이썬 리스트로 나누는 방법을 사용할 수 있다. 하지만 사실 이 과정은 그리 간단하지 않다.

여기에서는 훈련 데이터에서 첫 번째 문장을 공백을 기준으로 나누는 예시를 보겠다:

```python
# 첫 텍스트 요소를 스트링으로 저장
first_text = train.text.values[0]
print(first_text)
print('='*90)
print(first_text.split(' '))
```

하지만 첫 번째 토큰화 시도를 보면, 문장을 개별적인 요소들(혹은 용어)로 나누는 과정이 완전히 정확하지 않다는 것을 알 수 있다. 예를 들어, 리스트의 두 번째 요소인 "process,"를 보자. 쉼표(구두점)가 포함되어 있으며, 이는 "process"와 함께 하나의 용어로 취급되고 있다. 이상적으로는 쉼표와 단어가 두 개의 다른 요소로 리스트에 나뉘어야 한다. 순수 파이썬 리스트 연산만으로 이 작업을 수행하려면 꽤 복잡할 것이다. 그래서 이 부분에서 NLTK 라이브러리가 유용하게 사용된다. NLTK에는 단어와 구두점을 자동으로 분리하여 개별 요소로 나누는 "word_tokenize()" (TreebankWord tokenizer)라는 편리한 메서드가 있다. 이를 사용하면 다음과 같이 처리할 수 있다:

```python
nltk.download('punkt')
```

```python
first_text_list = nltk.word_tokenize(first_text)
print(first_text_list)
```

### 2c. 형태소 분석과 표제어 분석

불용어 제거 후, 소개하고 싶은 NLP의 다음 단계는 형태소 분석의 과정이다. 이 단계에서의 작업은 유사한 단어들의 다양한 형태를 하나의 용어로 축소하려는 시도를 한다. (다양한 형태가 모두 하나의 단어 어간으로 축소된다). 예를 들어, "running", "runs", "run"이 있다면, 이 세 개의 서로 다른 단어들이 모두 "run"이라는 단어로 합쳐지기를 원할 것이다. (하지만 물론 과거, 현재, 미래 시제에 대한 세부 정보는 잃게 된다).

다시 NLTK를 사용할 수 있는데, NLTK는 Porter 어간 추출 알고리즘, Lancaster 어간 추출기, Snowball 어간 추출기 등 다양한 형태의 어간 추출기를 제공한다. 다음 예시에서는 Porter 어간 추출기 인스턴스를 생성하는 방법을 보여준다:

```python
stemmer = nltk.stem.PorterStemmer()
```

이제 이 테스트 단어('running', 'runs', 'run')을 가지고 형태소 분석기가 한 가지 형태소 단어로 축소하는지 확인해볼 수 있다. 편하게 다음과 같은 코드로 즉석에서 형태소 분석기를 테스트할 수 있다:

```python
print('The stemmed form of running is: {}'.format(stemmer.stem('running')))
print('The stemmed form of runs is: {}'.format(stemmer.stem('runs')))
print('The stemmed form of run is: {}'.format(stemmer.stem('run')))
```

보시다시피, 어간 추출(stemming)은 주어진 단어들을 기본 형태로 축소하는 데 성공했으며, 이는 학습 및 분류 작업을 수행할 때 단어 데이터셋의 크기를 줄이는 데 도움이 된다.

하지만 어간 추출에는 한 가지 결함이 있다. 그 과정은 단어의 끝부분을 잘라내는 꽤 거친 휴리스틱을 사용하여 특정 단어를 인간이 인식할 수 있는 기본 형태로 축소하려고 한다. 따라서 이 과정은 단어를 축소할 때 어휘나 단어 형태를 고려하지 않게 된다. 이 예시에서 보듯이:

```python
print('The stemmed form of leaves is: {}'.format(stemmer.stem('leaves')))
```

### 어간 추출 대신 표제어 추출

따라서 어간 추출 대신 사용할 수 있는 또 다른 방법은 표제어 추출(lemmatization)이다. 이 방법은 어간 추출과 동일한 효과를 목표로 한다. 그러나 어간 추출기와 달리, 표제어 추출은 실제 사전이나 어휘(표제어)를 기반으로 단어를 축소하기 때문에 의미가 없는 형태로 단어를 잘라내지 않는다. 여기에서는 NLTK를 다시 활용하여 표제어 추출기(WordNet 변형)를 초기화하고, 단어가 어떻게 축소되는지 확인할 수 있다:

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
lemm = WordNetLemmatizer()
print('The lemmatized form of leaves is: {}'.format(lemm.lemmatize('leaves')))
```

훌륭하다. 이제 표제어 추출기가 작동하고 훨씬 더 어휘적인 의미를 갖는 형태로 단어를 축소하는 것을 확인할 수 있다.

### 2d. 원시 텍스트 벡터화

NLP 문헌에서는 원시 텍스트를 분석하는 다양한 목적이 있다. 예를 들어, 하나의 텍스트와 다른 텍스트 간의 유사성을 비교하거나 (클러스터링 기법/거리 측정), 텍스트 분류 (이 대회의 목적) 또는 텍스트를 구성하는 주제를 밝혀내는 목적(이 노트북의 목표) 등이 있다. 텍스트의 주제를 밝혀내는 목적을 염두에 두고 이제 우리는 원시 텍스트를 기계 학습 모델에 어떻게 입력할지 생각해야 한다. 토큰화, 불용어 제거, 어간 추출(또는 표제어 추출)을 이미 다룬 후, 우리는 시작할 때보다 상대적으로 더 깨끗한 텍스트 데이터셋을 얻었다. 그러나 현재 시점에서, 원시 텍스트는 인간이 읽을 수는 있지만 여전히 기계가 읽을 수 있는 형태는 아니다. 기계는 숫자와 비트(bit)를 읽을 수 있기 때문에 먼저 텍스트를 숫자로 변환해야 한다. 이를 위해 매우 일반적인 접근법인 Bag-of-Words 방법을 활용한다.

### Bag of Words 접근법
이 접근법은 단어의 빈도를 시작점으로 사용하고, 각 단어가 문서 내에서 발생하는 횟수를 기록하는 벡터를 생성한다. 예를 들어, "I love to eat Burgers"와 "I love to eat Fries"라는 두 문장이 있을 때, 먼저 토큰화를 통해 6개의 단어로 이루어진 어휘를 얻을 수 있다 - [I, love, to, eat, Burgers, Fries].

Bag of Words 접근법을 사용하여 텍스트를 벡터화하면, 각 단어마다 하나의 고유한 벡터가 생성된다. 이제 숫자로 이루어진 행이 있으므로, 무엇이 열(혹은 특성)을 형성하는지 궁금할 것이다. 각 단어가 이제 이 변환된 데이터셋에서 개별적인 특성/열이 된다. 이를 설명하기 위해, Scikit-learn 라이브러리를 사용하여 단어 빈도(용어 빈도)를 생성하는 벡터를 만드는 CountVectorizer 메서드를 다음과 같이 구현할 수 있다:

```python
sentence = ['I love to eat Burgers', 'I love to eat Fries']
vectorizer = CountVectorizer(min_df=0)
sentence_transform = vectorizer.fit_transform(sentence)
```

### 벡터라이저를 데이터셋에 맞추기

여기에서는 CountVectorizer 함수를 통해 "vectorizer"라는 간단한 용어 빈도 객체를 초기화하고 생성한다. 내가 명시적으로 제공한 매개변수들(나머지는 기본값으로 두었다)은 최소한의 요구 사항이다. 여기서 "min_df"라는 매개변수는 최소 문서 빈도를 의미하며, 벡터라이저는 이 값보다 적게 등장하는 단어는 모두 제거한다(정수 또는 비율 형식으로 설정할 수 있다). 이 방법과 사용 가능한 나머지 매개변수들에 대한 자세한 정보는 Sklearn 웹사이트에서 확인할 수 있다.

마지막으로 fit_transform 메서드를 적용하는데, 이 메서드는 사실 두 단계로 나누어져 있다. 첫 번째 단계는 fit 메서드로, 벡터라이저가 제공된 데이터셋에 맞춰지며, 두 번째 단계는 transform 메서드로, 원시 텍스트가 벡터 형식으로 변환된다:

```python
print('The features are:\n {}'.format(vectorizer.get_feature_names()))
print('\nThe vectorized array looks like::\n {}'.format(sentence_transform.toarray()))
```

## 3. 주제 모델링

*Final Destination*에 도달했다(말장난을 좀 하자면). 이제 두 가지 다른 주제 모델링 기법을 구현할 것이다:

1. Latent Dirichlet Allocation (LDA) - 확률적 생성 모델로, 데이터셋에 내재된 주제를 밝혀내기 위해 단어에 가중치를 할당한다. 각 주제는 각 단어에 대해 다른 확률적 가중치를 할당한다.


2. Non-negative Matrix Factorization (NMF) - 입력 행렬을 두 개의 다른 행렬로 근사하는 방법으로, 행렬 내 값들이 모두 음수가 아니어야 한다는 제약이 있다.

```python
# 인기 단어들을 출력하기 위한 헬퍼 함수 정의
def print_top_words(model, feature_names, n_top_words):
    for index, topic in enumerate(model.components_):
        message = '\nTopic #{}:'.format(index)
        message += ' '.join([feature_names[i] for i in topic.argsort()[:-n_top_words -1 : -1]])
        print(message)
        print('='*70)
```

### 3a. 모든 전처리 단계를 하나로 묶기

이제 이전 섹션에서 언급한 모든 텍스트 전처리 단계를 하나로 묶는 절호의 기회이다. 여러분은 "우리가 정말로 다시 토큰화, 불용어 제거, 어간 추출/표제어 추출 등을 다시 정의해야 할까요?"라고 묻고 있을 것이다.

다행히도 우리는 그 과정을 다시 반복할 필요가 없다. 나는 Scikit-learn 벡터라이저에서 중요한 세부 사항을 생략했지만, 이 시점에서 언급할 것이다. CountVectorizer로 원시 텍스트를 벡터화할 때, 토큰화와 불용어 필터링의 두 단계가 자동으로 고급 구성 요소로 포함된다. 여기서, 앞서 2a 섹션에서 소개한 NLTK 토크나이저와는 달리, Scikit-learn의 토크나이저는 'a', 'w'와 같은 한 글자 단어를 자동으로 제거하고, 기본적으로 모든 단어를 소문자로 변환한다. Scikit-learn에서 불용어를 필터링하는 것은 'stop_words' 인자에 'english'를 입력하면 자동으로 내장된 영어 불용어 목록이 사용된다.

불행히도, 벡터라이저에는 내장된 표제어 추출기가 없으므로, 우리는 몇 가지 선택 사항이 있다. 데이터를 벡터화하기 전에 매번 표제어 추출을 따로 구현하거나, Scikit-learn 구현을 확장하여 이 기능을 포함하는 방법이 있다. 다행히도 후자의 옵션이 있다. 우리는 CountVectorizer 클래스를 확장하여 "build_analyzer" 메서드를 재정의하고, 각 원시 텍스트 목록에 대해 표제어 추출기를 구현할 수 있다.

### CountVectorizer 클래스를 표제어 추출기와 함께 확장하기

```python
lemm = WordNetLemmatizer()
class LemmaCountVectorizer(CountVectorizer):
    def build_analyzer(self):
        analyzer = super(LemmaCountVectorizer, self).build_analyzer()
        return lambda doc: (lemm.lemmatize(w) for w in analyzer(doc))
```

여기서 우리는 객체 지향 프로그래밍(OOP)의 개념을 활용했다. 본질적으로 원래 Scikit-learn의 CountVectorizer 클래스를 상속받아 서브클래스를 만들고, "build_analyzer" 메서드를 재정의하여 원시 텍스트 매트릭스의 각 목록에 대해 표제어 추출기를 구현했다.

```python
# 전체 학습 텍스트를 리스트에 저장
text = list(train.text.values)
# 덮어쓴 Count Vectorizer 호출
tf_vectorizer = LemmaCountVectorizer(max_df=0.95,
                                    min_df=2,
                                    stop_words='english',
                                    decode_error='ignore')
tf = tf_vectorizer.fit_transform(text)
```

### 3b. 잠재 디리클레 할당(Latent Dirichlet Allocation, LDA)

마지막으로, 주제 모델링과 두 가지 비지도 학습 알고리즘의 구현을 다룬다. 첫 번째로 다룰 방법은 잠재 디리클레 할당 (LDA)이다. LDA 알고리즘은 여러 가지 구현이 있지만, 이 노트북에서는 Scikit-learn의 구현을 사용할 것이다. 또 다른 매우 잘 알려진 LDA 구현은 Radim Rehurek의 gensim이므로, 그것도 확인해 보길 바란다.

### 코퍼스 - 문서 - 단어 : 주제 생성
LDA에서 모델링 과정은 세 가지를 중심으로 진행된다: 텍스트 코퍼스, 문서 D의 집합, 문서에서 단어 W. 따라서 알고리즘은 다음과 같은 방식으로 이 코퍼스에서 K개의 주제를 찾아내려고 시도한다(다이어그램으로 설명됨).

![image.png](http://scikit-learn.org/stable/_images/lda_model_graph.png)

각 주제 κ는 디리클레 분포 βk를 통해 모델링된다.

각 문서 d는 또 다른 디리클레 분포 α에 의해 모델링된다.

이후 문서 d에 대해 다항 분포를 통해 주제를 생성하고, 그 주제와 관련된 단어들을 다른 다항 분포를 통해 생성한다.

(이미지 출처: http://scikit-learn.org/stable/modules/decomposition.html#latentdirichletallocation)

LDA 알고리즘은 먼저 문서를 주제의 혼합 모델로 모델링한다. 이 주제들로부터 단어들은 각 주제의 확률 분포를 기반으로 가중치를 부여받는다. 이러한 확률적 단어 할당 방식 덕분에 LDA 사용자는 특정 단어가 주제에 얼마나 속할지 확률적으로 알 수 있다. 이후 특정 주제에 할당된 단어들의 집합을 통해, 우리는 그 주제가 무엇을 의미하는지에 대한 통찰을 얻을 수 있다.

표준 LDA 모델에서는 몇 가지 중요한 매개변수가 있으며, 이들을 프로그램적으로 조정할 필요가 있다:

1. n_components: 모델에 지정할 주제의 개수


2. α 매개변수: 문서 주제 사전과 연결된 디리클레 매개변수


3. β 매개변수: 주제 단어 사전과 연결된 디리클레 매개변수

알고리즘을 호출하려면, Sklearn의 LatentDirichletAllocation 함수로 LDA 인스턴스를 생성한다. 다양한 매개변수들은 이상적인 값이 검증을 통해 얻어졌어야 한다. 이 예시에서는 n_components(즉, 주제 개수)의 최적 값은 KMeans + 잠재 의미 분석(Latent Semantic Analysis, LSA) 방식을 통해 찾았다(이 방식에 대한 자세한 내용은 [여기의 논문]() 참조). 이 방식은 KMeans 클러스터의 개수와 LSA 차원의 수를 반복하여 최고의 실루엣 평균 점수를 찾는 방법이다.

```python
lda = LatentDirichletAllocation(n_components=11, max_iter=5, learning_method='online',
                               learning_offset=50., random_state=0)
```

```python
lda.fit(tf)
```

### LDA로 생성된 주제들

우리는 이전에 정의한 보조 함수 "print_top_words"를 활용하여 LDA로 생성된 각 주제에 할당된 상위 10개 단어를 출력할 것이다. 주제의 수는 함수의 n_components 매개변수를 통해 조정된다.

```python
n_top_words = 40
print('\nTopics in LDA model: ')
tf_feature_names = tf_vectorizer.get_feature_names()
print_top_words(lda, tf_feature_names, n_top_words)
```

```python
first_topic = lda.components_[0]
second_topic = lda.components_[1]
third_topic = lda.components_[2]
fourth_topic = lda.components_[3]
```

```python
first_topic.shape
```

### 주제의 워드 클라우드 시각화

```python
first_topic_words = [tf_feature_names[i] for i in first_topic.argsort()[:-50 -1 : -1]]
second_topic_words = [tf_feature_names[i] for i in second_topic.argsort()[:-50-1:-1]]
third_topic_words = [tf_feature_names[i] for i in third_topic.argsort()[:-50-1:-1]]
fourth_topic_words = [tf_feature_names[i] for i in fourth_topic.argsort()[:50-1:-1]]
```

### 첫 주제에 대한 워드 클라우드

```python
# 카테고리 데이터프레임 아래의 값에 대한 워드 클라우드 생성
firstcloud = WordCloud(
    stopwords=STOPWORDS,
    background_color='black',
    width=2500,
    height=1800
).generate(' '.join(first_topic_words))
plt.imshow(firstcloud)
plt.axis('off')
plt.show()
```

```python
cloud = WordCloud(
    stopwords=STOPWORDS,
    background_color='black',
    width=2500,
    height=1800
).generate(' '.join(second_topic_words))
plt.imshow(cloud)
plt.axis('off')
plt.show()
```

```python
cloud = WordCloud(
    stopwords=STOPWORDS,
    background_color='black',
    width=2500,
    height=1800
).generate(' '.join(third_topic_words))
plt.imshow(cloud)
plt.axis('off')
plt.show()
```

```python
cloud = WordCloud(
    stopwords=STOPWORDS,
    background_color='black',
    width=2500,
    height=1800
).generate(' '.join(fourth_topic_words))
plt.imshow(cloud)
plt.axis('off')
plt.show()
```
